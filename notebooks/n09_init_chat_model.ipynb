```python
import os

from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
from langchain.tools import tool

# 1.加载环境变量
load_dotenv()  # 读取.env文件，加入系统环境变量中


# 2.定义工具，就是定义函数，给函数加装饰器 @tool
@tool
def get_weather(location: str):
    """查询指定location的天气"""
    return f"Current weather in {location} is sunny"


# 3.定义模型，默认初始化，只传名字即可，也可以加上模型自身的参数
model = init_chat_model(
    "qwen3.8-max",
    model_provider="openai",
    api_key=os.getenv("DASHSCOPE_API_KEY"),
    # base_url=os.getenv("DASHSCOPE_BASE_URL"),  # 报错，已经在.env里配置了
    base_url="https://dashscope.aliyuncs.com/compatible-mode/v1"  # 正常
)

# 4.定义Agent，绑定模型和工具
agent = create_agent(
    model=model,
    tools=[get_weather]
)

# 5.调用Agent，用invoke函数，传入一个Dict，字典要有messages属性，里面是消息的列表
response = agent.invoke({
    "messages": [{"role": "user", "content": "你是谁"}]
})

print(response)
```
