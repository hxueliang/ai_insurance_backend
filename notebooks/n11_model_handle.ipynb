```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

# 1.加载环境变量
load_dotenv()

# 2.初始化模型
model = init_chat_model(
    model="deepseek-v4-flash",
    extra_body={"thinking": {"type": "disabled"}}
)

"""
# 3.1调用模型，invoke调用，参数是str，也可以传list[dict]，返回值是一个 AIMessage
# 对比
# Agent.invoke调，参数是Dict，里面有messages列表，返回值是Dict，里面有messages列表
"""
response = model.invoke("你是谁")
print(response)
```

```python
"""
# 3.2调用模型，invoke调用，参数是list，返回值是一个 AIMessage
"""
response = model.invoke([
    {"role": "system", "content": "你扮演小团团，用小团团的语气来回答问题。"},
    {"role": "user", "content": "你是谁?"},
])
print(response)
```
